## Q2 — `for` loop &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Loop through the list below and print only the **even numbers**.

```python
numbers = [3, 8, 15, 22, 7, 40, 13, 6, 9, 18]
```

**Expected output:**

````python
# Q2 — Write your code here
numbers = [3, 8, 15, 22, 7, 40, 13, 6, 9, 18]

for i in numbers:
    if i % 2 == 0:
        print(i)
````

## Q3 — `while` loop                                                                    [20 marks]

Use a `while` loop to print the **Fibonacci series** up to 10 terms.

In Fibonacci series, each number is the sum of the two numbers before it.
Start with `a = 0` and `b = 1`.

**Expected output:**

0
1
1
2
3
5
8


````python
a = 0
b = 1
count = 0

while count < 7:
    print(a, end=' ')
    c = a + b       
    a = b           
    b = c
    count += 1
````

## Q4 — Function &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Write a function `is_even(n)` that returns `True` if a number is even, `False` if odd.  

Test with `is_even(4)`, `is_even(7)`, `is_even(0)`.  
**Expected output:**
```
4 is Even
7 is Odd
0 is Even
```

````python
# Q4 — Write your code here

def is_even(n):
    if n % 2 == 0:
        return f"{n} is Even" 
    else:
        return f"{n} is Odd" 
````

````python
is_even(4)
````

````python
is_even(7)
````

````python
is_even(0)
````

## Q5 — Combined: Function + for + if/else &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;[20 marks]

Write a function `check_pass(scores)` that loops through a list of marks and prints each score as `Pass` (≥ 50) or `Fail`.

Call with:
```python
scores = [72, 45, 88, 30, 55, 49, 91]
```
**Expected output:**
```
72 → Pass
45 → Fail
88 → Pass
30 → Fail
55 → Pass
49 → Fail
91 → Pass
```

````python
# Q5 — Write your code here
scores = [72, 45, 88, 30, 55, 49, 91]

def check_pass(scores):
    for i in scores:
        if i >= 50:
            print(f"{i} -> Pass")
        else:
            print(f"{i} -> Fail")
````

````python
check_pass([72, 45, 88, 30, 55, 49, 91])
````
