```python
# Question 1: Rotate a List K Times
# Problem Statement:
# Given a list of integers and an integer K, 
# rotate the list to the right by K positions.
# Example 1
# Input:
# nums = [1, 2, 3, 4, 5]
# k = 2
# Output:
# [4, 5, 1, 2, 3]
```

```python
nums = [1,2,3,4,5]
k = 2
store = 0

for i in range(k):
    store = nums[len(nums)-1]
    for j in range(len(nums)-1,-1,-1):
        nums[j] = nums[j-1]
    nums[0] = store

print(nums)
```

```python
# Question 2: Check Palindrome List Using Recursion
# Problem Statement:
# Write a recursive function to determine 
# whether a given list is a palindrome. 
# A list is called a palindrome if it
# reads the same forward and backward.
```

```python
def is_palindrome(s,*args):

    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False

    return is_palindrome(s[1:-1])

value = input()
if is_palindrome(value):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Question 4: Find the First Non-Repeating Element
# Problem Statement:
# Given a list of integers, find the first element that occurs only once. If no such element exists, return -1.
# Example 1
# Input:
# nums = [4, 5, 1, 2, 0, 4, 1, 2]
# Output:
# 5
```

```python
nums = [4, 5, 1, 2, 0, 4, 1, 2]

d = {}
for x in nums:
    if x in d:
        d[x] = d[x] + 1
    else:
        d[x] = 1

for x in d:
    if d[x] == 1:
        print(x)
        break
```

```python
# Question 5: Count Frequency of Each Element
# Problem Statement:
# Given a list of integers, count how many times each element 
# appears in the list and store the result in a
# dictionary.
```
