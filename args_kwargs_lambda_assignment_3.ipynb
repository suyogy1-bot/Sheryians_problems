# Q2. Sum of Even Numbers using *args

## Problem Statement
Write a Python function using *args to calculate the sum of all even numbers.

## Sample Function Call
```python
even_sum(1,2,3,4,5,6)
```

## Sample Output
```python
12
```


````python
def even_sum(*args):
    total = 0
    for i in args:
        if i % 2 == 0:
            total += i
    return total
````

````python
even_sum(1,2,3,4,5,6)
````

# Q3. User Profile Generator using **kwargs

```python
profile(name='Aman', city='Delhi', skill='Python')
```

## Sample Output

# Q6. Sort Dictionary by Value using Lambda

## Problem Statement
Write a Python program to sort a dictionary based on values using lambda.

## Sample Input
```python
{'a':3, 'b':1, 'c':2}
```

## Sample Output
```python
[('b',1), ('c',2), ('a',3)]
```


````python
key = lambda item: item[1]
````

````python
key(('a',3))
````

Step by step in your head:

- ‎`d.items()` → ‎`[('a', 3), ('b', 1), ('c', 2)]` (order may vary)

- ‎`sorted` takes each tuple, say ‎`('a', 3)`, and calls ‎`key(('a', 3))`.

- Inside the lambda: ‎`item[1]` → ‎`3`.

- It uses these numbers ‎`3, 1, 2` to decide order.



````python
dict = {'a':3, 'b':1, 'c':2}
print(dict.items()) # It will show tuples
````

````python
dict_sort = lambda d: sorted(d.items(), key=lambda item: item[1])
````

````python
dict_sort({'a':3, 'b':1, 'c':2})
````

# Q7. Filter Odd Numbers using Lambda

## Problem Statement
Write a Python program using lambda and filter() to extract odd numbers from a list.

## Sample Input
```python
[1,2,3,4,5,6,7]
```

## Sample Output
```python
[1,3,5,7]
```


````python
nums = [1,2,3,4,5,6,7]
odd = list(filter(lambda x: x % 2 != 0, nums))
````

Step-by-step in your head:

- ‎`filter` takes each element from ‎`nums` one by one (so ‎`x` is 1, then 2, then 3, …).

- The lambda returns ‎`True` if ‎`x % 2 != 0` (odd), ‎`False` otherwise.

- ‎`filter` keeps only the elements where lambda returned ‎`True`, which gives you the odd numbers.

````python
print(odd)
````

# Q8. Convert Strings to Uppercase using Lambda

## Problem Statement
Write a Python program using map() and lambda to convert all strings into uppercase.

## Sample Input
```python
['python', 'data', 'science']
```

## Sample Output
```python
['PYTHON', 'DATA', 'SCIENCE']
```


````python
l = ["python","DAt_a", "science"]
upper = map(lambda x: x.upper(), l)
````

````python
print(list(upper)) # Transformed Data
````

### Why `map()` output needs transformation before printing

- `map()` does **not** return a list; it returns a **map object**, which is an **iterator**.  
  Printing the iterator directly shows something like `<map object at 0x...>` instead of the values.

- The iterator uses **lazy evaluation**: it doesn’t compute results immediately.  
  Values are generated only when you iterate over it, which saves memory (e.g., for very large input lists).

- Because the results are not stored in a concrete container, Python cannot display all of them when you just `print(squared)`.

- To actually see the values, you must **consume** the iterator, for example:
  - `list(squared)` → list of results  
  - `tuple(squared)` → tuple of results  
  - `print(*map(func, iterable))` → unpacked values  
  - `for x in squared: print(x)` → iterate and print

- **Important:** Iterators are **one-time use**.  
  Once you convert a `map` object to a list or loop over it, it becomes empty, and further iterations produce no values.


## Core Comparison

| Feature | `map(function, iterable)` | `filter(function, iterable)` |
| --- | --- | --- |
| Primary Goal | Transforms elements. | Selects elements. |
| Lambda Output | Can return any value (numbers, strings, objects). | Must return a boolean (`True` or `False`). |
| Output Size | Always the same length as the original input. | Shorter or equal length compared to the input. |


# Q9. Student Marks Analyzer using *args

## Problem Statement
Write a Python function using *args to calculate:
- Total Marks
- Average Marks
- Highest Marks

## Sample Function Call
```python
marks(70,80,90,60)
```

## Sample Output
```python
Total: 300
Average: 75.0
Highest: 90
```


````python
def marks(*args):
    Total = sum(args)
    Average = sum(args) / len(args)
    Highest = max(args)
    print(f"\nTotal: {Total}\nAverage: {Average}\nHighest: {Highest}")
````

````python
marks(70,80,90,60)
````

# Q10. Dynamic Bill Generator using **kwargs

## Problem Statement
Write a Python function using **kwargs to print item names and prices, and calculate total bill amount.

## Sample Function Call
```python
bill(Shirt=1200, Shoes=2500, Watch=3000)
```

## Sample Output
```python
Shirt : 1200
Shoes : 2500
Watch : 3000
Total Bill : 6700
```


````python
def bill(**kwargs):
    total = 0
    for key, value in kwargs.items():
        print(f"{key} : {value}")
        total += value
    print(f"Total Bill : {total}")
````

````python
bill(Shirt=1200, Shoes=2500, Watch=3000)
````
