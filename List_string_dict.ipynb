## 1. Reverse a List Without Using reverse()

### Problem Statement
Write a program to reverse a list without using the built-in reverse() method.

### Example Input
```python
1 2 3 4 5
```

### Example Output
```python
[5, 4, 3, 2, 1]
```


````python
# Method 1 - Slicing 
l = [1,2,3,4,5]
l[::-1]
````

````python
l = [1,2,3,4,5]
for i in l:
    print(i)
````

````python
for i in range(len(l)):
    print(i)
````

````python
for i in range(len(l)):
    print(l[i])
````

````python
# Method 2 - Swapping

l = [1,2,3,4,5]
a = len(l)-1

for i in range(len(l)//2):
    l[i],l[a] = l[a],l[i] 
    i = i + 1
    a = a - 1

print(l)
````


## 2. Find Duplicate Elements in a List

### Problem Statement
Write a program to find all duplicate elements in a list.

### Example Input
```python
1 2 3 2 4 5 1
```

### Example Output
```python
[1, 2]
```


````python
string = "12234456666"

numbers_int = [int(i) for i in string]

seen = []
duplicates = []

for number in numbers_int:
    if number in seen and number not in duplicates:
        duplicates.append(number)
    elif number not in seen:
        seen.append(number)

print(duplicates)
````

````python
string = "1 2 3 2 4 5 1"
numbers = string.split()

numbers_int = [int(i) for i in numbers]

seen = []
duplicates = []

for number in numbers_int:
    if number in seen and number not in duplicates:
        duplicates.append(number)
    elif number not in seen:
        seen.append(number)

print(duplicates)
````


## 3. Count Frequency of Characters

### Problem Statement
Write a program to count the frequency of each character in a string.

### Example Input
```python
hello
```

### Example Output
```python
{'h':1, 'e':1, 'l':2, 'o':1}
```


````python
string = "Hello".lower()

freq = {}

# Ternary method
for i in string: freq[i] = freq[i] + 1 if i in freq else 1
print(freq)
````

````python
# Multiple lines - easier to read

string = "hello"
freq = {}

for i in string:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)
````

````python
from collections import Counter

string = "hello"
freq = Counter(string)

print(freq)
````

````python
freq = {'h': 1, 'e': 1, 'l': 2, 'o': 1}

sorted(freq.items())
````

````python
freq = {'h': 1, 'e': 1, 'l': 2, 'o': 1}

# Sort by key A->Z
sort = dict(sorted(freq.items()))
print(sort)
````

````python
# Sort by value low->high
sort = dict(sorted(freq.items(), key=lambda x: x[1]))
print(sort)
````

````python
# Sort by value high->low
sort = dict(sorted(freq.items(), key=lambda x: x[1], reverse=True))
print(sort)
````


## 4. Check Palindrome String

### Problem Statement
Write a program to check whether a string is palindrome or not.

### Example Input
```python
madam
```

### Example Output
```python
Palindrome
```


````python
string = "madam"
rev_str = string[::-1]

if string == rev_str:
    print("Palindrome")
else:
    print("Not Palindrome")
````
