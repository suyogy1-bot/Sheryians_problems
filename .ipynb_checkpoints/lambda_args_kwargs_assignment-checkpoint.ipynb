## 1. Print All Values using *args

### Problem Statement
Write a function using *args to print all values passed to the function.

### Example Input
```python
10, 20, 30
```

### Example Output
```python
10 20 30
```


````python
def argprint(*args):
    print(args)
    print(*args)
````

````python
argprint(10,20,30)
````


## 4. Print Student Details using **kwargs

### Problem Statement
Write a function using **kwargs to print student details.

### Example Input
```python
name='Parth', age=22
```

### Example Output
```python
name : Parht
age : 22
```


````python
def student(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
````

````python
student(name = "Parth", age = 22)
````


## 5. Print Keys using **kwargs

### Problem Statement
Write a function using **kwargs to print all keys.

### Example Input
```python
city='Bhopal', state='MP'
```

### Example Output
```python
city
state
```


````python
def print_key(**kwargs):
    for key in kwargs.keys():
        print(key)
````

````python
print_key(city = 'Bhopal', state="MP")
````


## 6. Print Values using **kwargs

### Problem Statement
Write a function using **kwargs to print all values.

### Example Input
```python
a=10, b=20
```

### Example Output
```python
10
20
```


````python
def print_value(**kwargs):
    for value in kwargs.values():
        print(value)
````

````python
print_value(a=10, b=20)
````


## 13. Check Even or Odd using Lambda

### Problem Statement
Write a lambda function to check whether a number is even or odd.

### Example Input
```python
8
```

### Example Output
```python
Even
```


````python
even_odd = lambda x : "Even" if x % 2 == 0 else "Odd"
````

````python
print(even_odd(8))
````


## 14. Find Greater Number using Lambda

### Problem Statement
Write a lambda function to find greater number between two values.

### Example Input
```python
15, 25
```

### Example Output
```python

```


````python
greater = lambda a,b : a if a > b else b
````

````python
print(greater(14,25))
````
