# Q1. Reverse Each Word in a String

## Problem Statement
Write a Python program to take a sentence from the user and reverse each word individually.

## Sample Input
```python
Enter sentence: hello world
```

## Sample Output
```python
olleh dlrow
```


````python
sentence = input("Enter sentence: ")
words = sentence.split()
reverse_list = []

for i in words:
    word = i[::-1]
    reverse_list.append(word)

result = " ".join(reverse_list)
print(result)
````

# Q2. Find Frequency of Characters

## Problem Statement
Write a Python program to count the frequency of each character in a string.

## Sample Input
```python
Enter string: apple
```

## Sample Output
```python
a : 1
p : 2
l : 1
e : 1
```


````python
string = input("Enter string: ")
d = {}

for i in string:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1

for key,value in d.items():
    print(f"{key} : {value}")
````

# Q3. Remove Duplicate Elements from List

## Problem Statement
Write a Python program to remove duplicate elements from a list.

## Sample Input
```python
Enter list: [1,2,2,3,4,4,5]
```

## Sample Output
```python
[1, 2, 3, 4, 5]
```


````python
user_input = input("Enter list: ")
numbers = eval(user_input)
new_list = []

for i in numbers:
    if i not in new_list:
        new_list.append(i)

print(new_list)
````

# Q4. Find Second Largest Number in List

## Problem Statement
Write a Python program to find the second largest element in a list.

## Sample Input
```python
Enter list: [10, 20, 5, 40, 30]
```

## Sample Output
```python
30
```


````python
user_input = input("Enter list: ")
number = eval(user_input)
largest = number[0]
sec_largest = 0

for i in number:
    if i > largest:
        sec_largest = largest
        largest = i
    elif i > sec_largest and i != largest:
        sec_largest = i

print(sec_largest)    
````

# Q5. Merge Two Dictionaries

## Problem Statement
Write a Python program to merge two dictionaries into one.

## Sample Input
```python
dict1 = {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}
```

## Sample Output
```python
{'a': 1, 'b': 2, 'c': 3, 'd': 4}
```


````python
dict1= {'a':1, 'b':2}
dict2 = {'c':3, 'd':4}

new_dict = {}

for key, value in dict1.items():
    new_dict[key] = value

for key, value in dict2.items():
    new_dict[key] = value

print(new_dict)
````

````python
# Q5. Merge Two Dictionaries

dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

merged = {}

def add_to_dict(d, key, value):
    # if key already exists
    if key in d:
        # if it's already a list, just append
        if isinstance(d[key], list):
            d[key].append(value)
        else:
            # convert existing single value to a list
            d[key] = [d[key], value]
    else:
        # key not present yet, store single value
        d[key] = value

for key, value in dict1.items():
    add_to_dict(merged, key, value)

for key, value in dict2.items():
    add_to_dict(merged, key, value)

print(merged)
````
